### **SGD Rgressor**

```python
from sklearn.datasets import load_diabetes

import numpy as np
```

```python
X,y = load_diabetes(return_X_y = True)
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2, random_state = 2)
```

```python
from sklearn.linear_model import SGDRegressor

reg = SGDRegressor(penalty = 'l2',max_iter = 500, eta0 = 0.1,learning_rate = 'constant',alpha = 0.001)
```

```python
reg.fit(X_train,y_train)
```

```python
print("Coefficients :",reg.coef_)
print("Intercept :",reg.intercept_)
```

```python
y_pred = reg.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("Mean Squared Error :",mean_squared_error(y_test,y_pred))
print("R2 Score :",r2_score(y_test,y_pred))
```

### **Ridge Regression Using Gradient Descent**

```python
from sklearn.linear_model import Ridge

reg1= Ridge(alpha = 0.001,max_iter = 500, solver = 'sparse_cg')
```

```python
reg1.fit(X_train,y_train)
```

```python
print("Coefficients :",reg1.coef_)
print("Intercept :",reg1.intercept_)
```

```python
y_pred = reg1.predict(X_test)
```

```python
print("Mean Squared Error :",mean_squared_error(y_test,y_pred))
print("R2 Score :",r2_score(y_test,y_pred))
```

```python
# creating class
class SGD_Ridge:

  def __init__(self,epochs,learning_rate,alpha):
    # Constructor to initialize hyperparameters
    self.epochs = epochs  # Number of iterations (epochs) for gradient descent
    self.learning_rate = learning_rate # Step size for updating weights
    self.alpha = alpha # Regularization strength (L2 penalty)
    self.coef_ = None # Model coefficients (weights for features)
    self.intercept_ = None # Model intercept (bias term)

  def fit(self,X_train,y_train):
    # Initialize intercept and coefficients before training
    self.intercept_ = 0
    # Coefficients for each feature, initialized to ones
    self.coef_ = np.ones(X_train.shape[1])
    # Combine intercept and coefficients into a single weight vector
    # The first element is the intercept, followed by feature coefficients
    weight = np.insert(self.coef_,0,self.intercept_)

    # Add a bias term (column of ones) to X_train for intercept calculation
    # This allows treating the intercept as another coefficient in the linear model
    X_train = np.insert(X_train,0,1,axis = 1)


    # Perform Batch Gradient Descent for a specified number of epochs
    for i in range(self.epochs):
      # Calculate the gradient of the Ridge regression cost function
      # The full gradient for Ridge regression
      # (X_train.T @ (X_train @ weight - y_train) + alpha * weight

      # Term related to the data (mean squared error part)
      # np.dot(X_train.T,X_train).dot(weight) - np.dot(X_train.T,y_train) is equivalent to X_train.T @ (X_train @ weight - y_train)
      data_gradient_term = np.dot(X_train.T,X_train).dot(weight) - np.dot(X_train.T,y_train)

      # Term related to regularization (L2 penalty)
      regularization_gradient_term = (self.alpha * weight)

      # Combine the terms and scale the data-dependent term by 1/n_samples
      # This scaling is crucial to prevent numerical overflow and ensure stable learning
      weight_der = data_gradient_term + regularization_gradient_term

      # Update weights by taking a step in the opposite direction of the gradient
      weight = weight - self.learning_rate * weight_der

    # Separate intercept and coefficients after training is complete
    self.intercept_  = weight[0] # The first element of the final weight vector is the intercept
    self.coef_ = weight[1:] # The remaining elements are the coefficients for each feature

  def predict(self, X_test):
    # Make predictions on new data using the learned coefficients and intercept

    return np.dot(X_test,self.coef_) +  self.intercept_
```

```python
sgd_reg = SGD_Ridge(epochs = 500,learning_rate = 0.005,alpha = 0.001)
```

```python
sgd_reg.fit(X_train,y_train)
```

```python
print("Coefficients :",sgd_reg.coef_)
print("Intercept :",sgd_reg.intercept_)
```

```python
pred = sgd_reg.predict(X_test)
```

```python
print("Mean Squared Error :",mean_squared_error(y_test,pred))
print("R2 Score :",r2_score(y_test,pred))
```
